Simulating the fraud detection problem from Question 1 in Part 1.

```python
import numpy as np
import pandas as pd

# Step 1: Set random seed for reproducibility
np.random.seed(42)

# Step 2: Create 100,000 transactions
n_transactions = 100000

# Step 3: Determine which transactions are fraudulent (0.1% rate)
# Hint: Use np.random.rand() < 0.001
is_fraud = np.random.rand(n_transactions) < 0.001

# Step 4: Apply fraud detection system
# For fraudulent transactions: 98% chance of being flagged
# For legitimate transactions: 5% chance of being flagged (false positive)

flagged = np.zeros(n_transactions, dtype=bool)

# Use np.random.rand() to determine if each transaction gets flagged
# You may use Boolean indexing or a loop.
r = np.random.rand(n_transactions)
flagged[is_fraud] = r[is_fraud] < 0.98
flagged[~is_fraud] = r[~is_fraud] < 0.05

# Step 5: Calculate empirical probability
# P(Fraud | Flagged) = (Number of flagged AND fraud) / (Total flagged)

fraud_and_flagged = np.sum(is_fraud & flagged)
total_flagged = np.sum(flagged)
empirical_prob = fraud_and_flagged / total_flagged

print(f"Empirical P(Fraud | Flagged) = {empirical_prob:.4f}")
print(f"Theoretical P(Fraud | Flagged) = {0.00098 / 0.05093:.4f}")

# Step 6: Create confusion matrix (counts)
true_positive = np.sum(is_fraud & flagged)  # Fraud AND Flagged
false_positive = np.sum(~is_fraud & flagged)  # Not Fraud AND Flagged
true_negative = np.sum(~is_fraud & ~flagged)  # Not Fraud AND Not Flagged
false_negative = np.sum(is_fraud & ~flagged)  # Fraud AND Not Flagged

confusion_matrix = pd.DataFrame(
    {"Flagged": [true_positive, false_positive],
     "Not Flagged": [false_negative, true_negative]},
    index=["Fraud", "Legitimate"],
)
confusion_matrix
```

The empirical P(Fraud | Flagged) = 0.0200 is close to the 0.0192 from 1(a). In this sample there are 102 fraud transactions and 100 of them got flagged, but 4,892 legitimate transactions also got flagged, so only 100 out of 4,992 flags are real fraud. It's not exactly 0.0192 because of randomness, we expected about 100 frauds and 4,995 false positives.

With specificity 0.99 the false positives drop from 4,892 to 974 and P(Fraud | Flagged) goes up to about 0.093, close to the 0.0893 from 1(b).
