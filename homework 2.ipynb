1\. Utility from leisure: $U(h)=20h-0.5h^2$, h = hours of leisure per day

(a) marginal utility = derivative of U with respect to h (power rule):

$$
U'(h)=20(1)-0.5(2h)=20-h
$$

(b) set marginal utility equal to 10:

$$
20-h=10
$$

$$
h=20-10=10
$$

So marginal utility is 10 when the worker has 10 hours of leisure per day.

3\. Concavity / convexity of the production functions

rule from class (second derivative), with L > 0:

$Q''(L)<0$: concave, the marginal product gets smaller as L goes up, so decreasing returns

$Q''(L)>0$: convex, the marginal product gets bigger as L goes up, so increasing returns

$Q''(L)=0$: the slope doesn't change, so constant returns

(a) $Q(L)=10\sqrt{L}=10L^{1/2}$

first and second derivatives:

$$
Q'(L)=10\cdot\frac{1}{2}L^{-1/2}=5L^{-1/2}=\frac{5}{\sqrt{L}}
$$

$$
Q''(L)=5\cdot\left(-\frac{1}{2}\right)L^{-3/2}=-\frac{5}{2}L^{-3/2}=-\frac{5}{2L^{3/2}}
$$

returns: $Q'(L)>0$ but $Q''(L)<0$ for every L > 0, so Q is concave and has decreasing returns. Each extra unit of labor adds less output than the one before, e.g. $Q'(1)=5$ but $Q'(4)=5/2=2.5$.

(b) $Q(L)=L^2$

first and second derivatives:

$$
Q'(L)=2L,\quad Q''(L)=2
$$

returns: $Q''(L)=2>0$, so Q is convex and has increasing returns. The marginal product keeps going up, e.g. $Q'(1)=2$ and $Q'(2)=4$.

(c) $Q(L)=\ln(L)$

first and second derivatives:

$$
Q'(L)=\frac{1}{L}=L^{-1},\quad Q''(L)=-1\cdot L^{-2}=-\frac{1}{L^2}
$$

returns: $Q''(L)<0$ for every L > 0, so Q is concave and has decreasing returns, e.g. $Q'(1)=1$ but $Q'(2)=0.5$.

None of the three has $Q''(L)=0$, so none of them shows constant returns.

5\. $g(x,y)=2x^2+y^2-4xy+6x-2y+10$

(a) first-order conditions, set both partials equal to 0:

$$
g_x=4x-4y+6=0
$$

$$
g_y=2y-4x-2=0
$$

from $g_y=0$: 2y = 4x + 2, so y = 2x + 1

plug into $g_x=0$:

$$
4x-4(2x+1)+6=0
$$

$$
4x-8x-4+6=0
$$

$$
-4x+2=0\quad\text{so}\quad x=\frac{1}{2}
$$

$$
y=2\left(\frac{1}{2}\right)+1=2
$$

So there is only one critical point, (x, y) = (1/2, 2). The value of g there:

$$
g(0.5,\,2)=2(0.25)+2^2-4(0.5)(2)+6(0.5)-2(2)+10
$$

$$
=0.5+4-4+3-4+10=9.5
$$

(b) second-order partials:

$$
g_{xx}=4,\quad g_{yy}=2,\quad g_{xy}=g_{yx}=-4
$$

Hessian at (1/2, 2) (all the entries are constants, so it's the same at every point):

$$
H=\begin{bmatrix}g_{xx}&g_{xy}\\g_{yx}&g_{yy}\end{bmatrix}=\begin{bmatrix}4&-4\\-4&2\end{bmatrix}
$$

$$
D=\det(H)=g_{xx}g_{yy}-(g_{xy})^2=4(2)-(-4)^2=8-16=-8
$$

D < 0, so (1/2, 2) is a saddle point (iii), not a local max or a local min. Even though $g_{xx}=4>0$ and $g_{yy}=2>0$, the cross partial is big enough to make D negative, so g goes up in some directions and down in others.

# Part B. Python Practice

Find the exact area under $f(x)=x^2$ from 0 to 2.

1\. manual calculation, power rule $\int x^n\,dx=\frac{x^{n+1}}{n+1}+C$ with n = 2, then F(b) - F(a):

$$
\int_0^2 x^2\,dx=\left[\frac{x^3}{3}\right]_0^2=\frac{2^3}{3}-\frac{0^3}{3}=\frac{8}{3}-0=\frac{8}{3}\approx2.666667
$$

2\. complete the code: the height of each rectangle is f(x) and its area is height times width.

```python
# Define the function f(x) = x^2
def f(x):
    return x**2

# Estimate integral using rectangles
def estimate_area(a, b, n):
    # a = start point, b = end point, n = number of rectangles

    width = (b - a) / n  # width of each rectangle
    total_area = 0

    for i in range(n):
        # x-coordinate of left side of rectangle
        x = a + i * width

        # Height of rectangle is f(x)
        height = f(x)

        # Add area of this rectangle
        rectangle_area = height * width
        total_area = total_area + rectangle_area

    return total_area

actual_area = 8/3
estimated = estimate_area(0, 2, 10)


print("Actual area:", f"{actual_area:.6f}")
print("Estimated area:", f"{estimated:.6f}")
```

With n = 10 the estimate is 2.28, lower than 8/3. check by hand: width = 2/10 = 0.2 and the left sides are x = 0, 0.2, 0.4, ..., 1.8, so

$$
\text{area}\approx0.2\left(0^2+0.2^2+0.4^2+\cdots+1.8^2\right)
$$

$$
=0.2(0.04)(0+1+4+9+\cdots+81)
$$

$$
=0.008(285)=2.28
$$

It comes out too low because we use the left side of each rectangle and $x^2$ is increasing on [0, 2], so every rectangle sits under the curve.

3\. try different n:

```python
# Question 3: try different n
for n in [10, 100, 1000, 10000, 100000]:
    estimated = estimate_area(0, 2, n)
    error = actual_area - estimated
    print(f"n = {n}:  estimated area = {estimated:.6f},  error = {error:.6f}")
```

Yes, the approximation gets better as n increases. More rectangles means each one is thinner, so there is less gap between the rectangles and the curve and the estimate gets closer to 8/3 = 2.666667. Each time n is 10 times bigger the error is about 10 times smaller (0.39, 0.04, 0.004, 0.0004, 0.00004). The estimate is always a little below 8/3 since we use left endpoints and $x^2$ is increasing.
